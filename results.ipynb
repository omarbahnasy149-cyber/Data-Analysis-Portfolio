```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Create Sample Dataset
data = {
    "Date": pd.date_range(start="2023-01-01", periods=5),
    "Product_Category": ["Electronics", "Clothing", "Home", "Electronics", "Clothing"],
    "Sales": [250, 100, 150, 300, 120],
    "Region": ["North", "South", "East", "West", "North"]
}

df = pd.DataFrame(data)

# 2. Category Sales Summary
category_sales = df.groupby("Product_Category")["Sales"].sum()
print("Category Sales Summary:")
print(category_sales)

# 3. Top Region
print("\nTop Performing Region:")
print(df.groupby("Region")["Sales"].sum().idxmax())
```
